1) Em um cercado existem vários patos e coelhos. Escreva uma função em linguagem Python que recebe como parâmetro o total de cabeças e o total de pés, e retorne dois valores: o total de patos e o total de coelhos que se encontram nesse cercado. (4,0 pontos) Teste essa função elaborando um script que vai receber do usuário dois valores. Para tanto, leve em consideração que os dois últimos dígitos de seu RA sejam iguais a Y.  Entrada de Dados:    Total de Cabeças:  Y = _________                                       Total de Pés: Y*3 = _________  Levando em consideração o resultado de seu script... Responda:  Total de Patos:___________________        Total de Coelhos: ___________________  


```python
def calcular_animais(cabecas, pes):
    coelhos = (pes - 2 * cabecas) // 2
    patos = cabecas - coelhos
    return patos, coelhos

try:
    total_cabecas = int(input("Digite o total de cabeças (Y): "))
    total_pes = int(input("Digite o total de pés (Y*3): "))
    
    patos, coelhos = calcular_animais(total_cabecas, total_pes)
    
    print(f"\nResultado do script:")
    print(f"Total de Patos: {patos}")
    print(f"Total de Coelhos: {coelhos}")
except ValueError:
    print("Insira apenas números inteiros válidos.")


```

Faça um script em Python que recebe como entrada um número positivo no formato de String. Esse script deve devolver dois valores: o total da soma de todos os dígitos desse número e o valor da multiplicação de todos esses mesmos dígitos. Por exemplo: Se a entrada for 1234, o script deve devolver como resultado: Soma = 10 e Multiplicação = 24.       (3,0 pontos) Passe o número completo de seu RA para o script, e responda:  RA = _____________________________  Multiplicação = _____________________  Soma = ____________________________ 

def calcular_digitos(numero_str):
    soma = 0
    multiplicacao = 1
    
    for digito in numero_str:
        num = int(digito)
        soma += num
        multiplicacao *= num
        
    return soma, multiplicacao


ra_entrada = "46"
resultado_soma, resultado_mult = calcular_digitos(ra_entrada)

print(f"Soma = {resultado_soma}")
print(f"Multiplicação = {resultado_mult}")


Reverso do número. Faça uma função que retorne o reverso de um número inteiro informado. Por exemplo: 127 -> 721.   ((use os números de seu RA para testar))  

```python
def reverso_numero(numero):
    numero_str = str(numero)
   
    reverso_str = numero_str[::-1]
    return int(reverso_str)
meu_ra = 46
resultado = reverso_numero(meu_ra)

print(f"Número original (RA): {meu_ra}")
print(f"Reverso do número: {resultado}")
```
